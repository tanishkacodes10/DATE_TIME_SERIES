```python
from datetime import datetime

dt = datetime(2026, 2, 15, 10, 30)

print("Year:", dt.year)
print("Month:", dt.month)
print("Day:", dt.day)
print("Hour:", dt.hour)
print("Minute:", dt.minute)
```

```python
from datetime import datetime

dt1 = datetime(2026, 2, 15, 10, 0)
dt2 = datetime(2026, 2, 15, 12, 30)

diff = dt2 - dt1

print("Difference:", diff)
print("Hours:", diff.seconds // 3600)
print("Minutes:", (diff.seconds % 3600) // 60)
```

```python
import pandas as pd

dates = pd.to_datetime([
    "2026-02-01", "2026-02-02", "2026-02-04",
    "2026-02-06", "2026-02-08"
])

sales = pd.Series(
    [100, 150, 120, 180, 200],
    index=dates,
    name="units_sold"
)
```

```python
print(sales.index)
print(type(sales.index))
```

```python
print(sales.loc["2026-02-04"])
```

```python
print("First observation:", sales.iloc[0])
print("Last observation:", sales.iloc[-1])
```

```python
result = sales.loc["2026-02-02":"2026-02-06"]

print(result)
```

```python
print("Total Sales:", sales.sum())
print("Average Sales:", sales.mean())
```

Section D

```python
import pandas as pd

dates = pd.date_range(
    "2026-02-01",
    periods=6,
    freq="D"
)

daily_sales = pd.Series(
    [100, 120, 150, 130, 180, 200],
    index=dates
)

print(daily_sales)
```

```python
df = pd.DataFrame({
    "Current Sales": daily_sales,
    "Previous Day Sales": daily_sales.shift(1)
})

print(df)
```

```python
df = pd.DataFrame({
    "Current Sales": daily_sales,
    "Next Day Sales": daily_sales.shift(-1)
})

print(df)
```

```python
change = daily_sales - daily_sales.shift(1)

print(change)
```

```python
shifted_sales = daily_sales.shift(2, freq="D")

print("Original Series:")
print(daily_sales)

print("\nShifted Series:")
print(shifted_sales)
```
